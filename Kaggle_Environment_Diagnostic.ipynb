Yeni private Kaggle notebook: GPU T4 ×2 ve Internet On. Run All kullan; yalnız üretilen ZIP’i gönder. Bu sürüm paket kurmaz, model indirmez ve kernel/serving/collective testi çalıştırmaz.

```python
from pathlib import Path
import subprocess, sys
URL = "https://github.com/nourey/moe-mechanistic-kaggle.git"
TAG = "env-v0.1.0"
repo = Path("/kaggle/working/moe-mechanistic-kaggle-" + TAG)
if repo.exists():
    raise RuntimeError("Bu klasör zaten var; mevcut çalışmayı silmeden yeni notebook/klasör kullan.")
subprocess.run(["git", "clone", "--branch", TAG, "--depth", "1", URL, str(repo)], check=True)
print("Runner commit:", subprocess.check_output(["git", "-C", str(repo), "rev-parse", "HEAD"], text=True).strip())
```

```python
import json
from IPython.display import display, FileLink
completed = subprocess.run([sys.executable, "-B", str(repo / "diagnose.py"), "--output-root", "/kaggle/working"], text=True, capture_output=True)
print(completed.stdout)
if completed.stderr:
    print(completed.stderr)
completed.check_returncode()
line = next(x for x in reversed(completed.stdout.splitlines()) if x.startswith("EVIDENCE_JSON:"))
receipt = json.loads(line.split(":", 1)[1])
print("Bana gönderilecek kanıt ZIP'i:", receipt["archive"])
display(FileLink(receipt["archive"]))
```
