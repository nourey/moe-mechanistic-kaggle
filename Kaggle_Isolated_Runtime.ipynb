```python
from pathlib import Path
import subprocess, sys, json
from IPython.display import display, FileLink
TAG = "runtime-v0.2.1"
repo = Path("/kaggle/working/moe-mechanistic-kaggle-" + TAG)
if repo.exists():
    raise RuntimeError("Önceki çalışmayı silme. Yeni notebook/çalışma klasörü kullan.")
subprocess.run(["git", "clone", "--branch", TAG, "--depth", "1", "https://github.com/nourey/moe-mechanistic-kaggle.git", str(repo)], check=True)
print("Runner commit:", subprocess.check_output(["git", "-C", str(repo), "rev-parse", "HEAD"], text=True).strip())
```

```python
lines = []
process = subprocess.Popen([sys.executable, "-B", str(repo / "prepare_runtime.py"), "--output-root", "/kaggle/working"], text=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, bufsize=1)
for line in process.stdout:
    print(line, end="", flush=True)
    lines.append(line)
if process.wait() != 0:
    raise RuntimeError("Runner durdu; yukarıdaki hatayı ve varsa kanıt ZIP'ini koru.")
line = next(x for x in reversed(lines) if x.startswith("EVIDENCE_JSON:"))
receipt = json.loads(line.split(":", 1)[1])
print("Kurulum/import sonucu:", receipt["runtime_preparation"])
print("Bana gönderilecek ZIP:", receipt["archive"])
display(FileLink(receipt["archive"]))
```
